```python
import requests
from bs4 import BeautifulSoup

search_query = input("Enter product category: ")
page_no = 1

while True:

    url = f"https://www.flipkart.com/search?q={search_query}&page={page_no}"

    print("\nURL:", url)

    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                      "AppleWebKit/537.36 (KHTML, like Gecko) "
                      "Chrome/120.0.0.0 Safari/537.36",
        "Accept-Language": "en-US,en;q=0.9"
    }

    response = requests.get(url, headers=headers)

    if response.status_code == 200:

        soup = BeautifulSoup(response.text, "html.parser")

        products = soup.find_all("a", href=True)

        count = 0

        for item in products:

            text = item.get_text(" ", strip=True)

            if "/p/" in item.get("href", "") and text:

                print("Product:", text[:100])

                # Find price and discount from product container
                parent = item.parent

                for i in range(5):
                    if parent:
                        parent_text = parent.get_text(" ", strip=True)

                        if "₹" in parent_text:
                            break

                        parent = parent.parent

                if parent:
                    data = parent.get_text(" ", strip=True)

                    # Extract price
                    price_start = data.find("₹")

                    if price_start != -1:
                        price = data[price_start:].split()[0]
                    else:
                        price = "Not available"

                    # Find discount
                    discount = "Not available"

                    words = data.split()

                    for word in words:
                        if "%" in word and "off" in data.lower():
                            discount = word + " off"
                            break

                    print("Price:", price)
                    print("Discount:", discount)
                    print("-" * 50)

                    count += 1

                if count == 10:
                    break

        if count == 0:
            print("No products found on this page.")
            break

        print("Page no:", page_no)

    else:
        print("Failed to fetch page.")

    page_no += 1
```

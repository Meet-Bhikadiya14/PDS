```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson
```

```python
n = 10
p = 0.5

x_binomial = np.arange(0, n + 1)
binomial_prob = binom.pmf(x_binomial, n, p)

print("Binomial Distribution:")

for x, prob in zip(x_binomial, binomial_prob):
    print(f"P(X = {x}) = {prob:.4f}")
```

```python
lam = 4

x_poisson = np.arange(0, 15)
poisson_prob = poisson.pmf(x_poisson, lam)

print("Poisson Distribution:")

for x, prob in zip(x_poisson, poisson_prob):
    print(f"P(X = {x}) = {prob:.4f}")
```

```python
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.bar(x_binomial, binomial_prob)
plt.title("Binomial Distribution")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")

plt.subplot(1, 2, 2)
plt.bar(x_poisson, poisson_prob)
plt.title("Poisson Distribution")
plt.xlabel("Number of Events")
plt.ylabel("Probability")

plt.tight_layout()
plt.savefig("binomial_poisson_distribution.png")
plt.show()
```
